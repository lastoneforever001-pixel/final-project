```python
import time
import sys

def typing_print(text, speed=0.05):
    for char in text:
        sys.stdout.write(char)
        sys.stdout.flush()
        time.sleep(speed)
    print()

# 1. The Build-up
typing_print("Connecting to your heart server...", 0.06)
time.sleep(1)
typing_print("Analyzing compatibility... [100% Match Found]", 0.04)
time.sleep(1)
typing_print("\nI have a critical question to ask you:\n", 0.08)

# 2. The Big Question
proposal_text = "If I am the code, will you be my compiler? Because without you, my life has syntax errors. ❤️"
typing_print(proposal_text, 0.06)
time.sleep(0.5)

# 3. The Interactive Input
response = input("\nWill you be my partner for life? (yes/no): ").strip().lower()

if response == "yes":
    print("\n" + " " * 12 + "✨ CRITICAL SUCCESS ✨")
    # Prints a visual heart
    heart = [
        "  ***     ***  ",
        " *    * *    * ",
        "*      *      *",
        "*   YES! FOREVER *",
        " *           * ",
        "  *         *  ",
        "    *     *    ",
        "      * *      ",
        "       *       "
    ]
    for line in heart:
        print(line.center(40))
        time.sleep(0.1)
    print("\nWelcome to our infinite loop of happiness! 🥰💍")
else:
    print("\nSystem Error: Heart Broken. 💔 Retrying affection routine anyway...")
```

```python
import sys
import time


def slow_type(text, speed=0.06):
    """Prints text slowly to create suspense and emotional impact."""
    for char in text:
        sys.stdout.write(char)
        sys.stdout.flush()
        time.sleep(speed)
    print()


def ask_question(prompt, expected_keyword):
    """Ensures she answers sweetly before moving to the next stage."""
    while True:
        answer = input(prompt).strip().lower()
        if expected_keyword in answer:
            return True
        else:
            slow_type(
                "\nProcessing response... ❤️  My heart won't accept that answer. Try again! 🥰"
            )


# === STAGE 1: THE INITIALIZATION ===
slow_type("Initializing Emotional_Connection.exe...", 0.08)
time.sleep(1)
slow_type("Scanning memory banks for the most beautiful soul... 🔎", 0.05)
time.sleep(1)
slow_type("MATCH FOUND: It's you. It has always been you. ✨\n", 0.06)
time.sleep(1.5)

# === STAGE 2: THE LOVELY QUESTIONS ===
slow_type(
    "Before I reveal what's deep in my source code, I need to ask you a few things..."
)
time.sleep(1)

# Question 1
ask_question(
    "\n1. Do you know that my day completely lights up the moment you smile? (yes/no): ",
    "yes",
)
slow_type(
    "   ↳ My database confirms it. Your smile is my absolute favorite view. 😍"
)
time.sleep(1.5)

# Question 2
ask_question(
    "\n2. Will you promise to let me hold your hand through every high and every low? (yes/no): ",
    "yes",
)
slow_type(
    "   ↳ Connection locked. I promise to never let go, no matter the weather. 🌧️☀️"
)
time.sleep(1.5)

# Question 3
ask_question(
    "\n3. Do you believe in destiny? Because meeting you felt like the perfect design. (yes/no): ",
    "yes",
)
slow_type(
    "   ↳ Me too. Out of billions of people, my heart chose exactly you. 🧭"
)
time.sleep(1.5)

# === STAGE 3: THE CONFESSION ===
slow_type("\nLoading deep_confession_subroutine...", 0.08)
time.sleep(2)

confession = (
    "\nEvery single day spent with you feels like a beautiful dream.\n"
    "You are the rhythm to my heartbeat, the laughter in my joy, and the calm in my chaos.\n"
    "I don't just want a temporary spark; I want a lifetime of quiet mornings, silly late-night talks,\n"
    "and a love that grows stronger with every single breath we take.\n"
)
slow_type(confession, 0.07)
time.sleep(2)

# === STAGE 4: THE ULTIMATE PROPOSAL ===
slow_type("=" * 60, 0.02)
slow_type(
    "💍 THE ULTIMATE QUESTION 💍", 0.1
)  # Slow down for maximum impact
slow_type("=" * 60, 0.02)
time.sleep(1)

final_prompt = (
    "\nWill you make me the luckiest person in the universe?\n"
    "Will you walk beside me, dream with me, and be my forever love? ❤️\n"
    "\nPlease type 'YES' to accept: "
)

ask_question(final_prompt, "yes")

# === STAGE 5: THE CELEBRATION ===
print("\n")
heart_frames = [
    "      ****     ****      ",
    "    ******** *********    ",
    "   *******************    ",
    "   *******************    ",
    "    *****************     ",
    "      *************       ",
    "        *********         ",
    "          *****           ",
    "            *             ",
]

for line in heart_frames:
    print(line.center(60))
    time.sleep(0.1)

slow_type("\n✨ PROPOSAL ACCEPTED! INFINITE HAPPINESS SUBROUTINE ACTIVATED! ✨")
slow_type("I love you more than all the stars in the night sky. 🥰🥂💍")
```
