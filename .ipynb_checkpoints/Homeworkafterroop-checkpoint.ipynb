```python

def bank_account():
    balance = 10000

    def deposit():
        nonlocal balance
        amount = int(input("Enter deposit amount: "))
        balance = balance + amount
        print("Amount deposited successfully")

    def withdraw_amount():
        nonlocal balance
        amount = int(input("Enter withdrawal amount: "))

        if amount <= balance:
            balance = balance - amount
            print("Withdrawal successful")
        else:
            print("Insufficient balance")

    def check_balance():
        print("Current balance:", balance)

    deposit()
    withdraw_amount()
    check_balance()


bank_account()

```

```python
class EncapsulationDemo:
    def __init__(self, value):
        self.__private_var = value  # Private variable

    def __private_method(self):  # Private method
        return f"Inside private method. Value is: {self.__private_var}"

    # Public method to access private variable and method
    def get_value(self):
        print(self.__private_method())
        return self.__private_var

    # Public method to modify private variable
    def set_value(self, new_value):
        self.__private_var = new_value
        print("Private variable updated successfully.")


# Testing the class
obj = EncapsulationDemo(10)
print("Initial Access:", obj.get_value())
obj.set_value(25)
print("Modified Access:", obj.get_value())
```

```python
class Point:
    def __init__(self, x, y):
        self.x = x
        self.y = y

    def __add__(self, other):
        return Point(self.x + other.x, self.y + other.y)

    def __sub__(self, other):
        return Point(self.x - other.x, self.y - other.y)

    def __str__(self):
        return f"Point({self.x}, {self.y})"


# Testing operator overloading
p1 = Point(2, 3)
p2 = Point(4, 5)
print("Addition:", p1 + p2)
print("Subtraction:", p2 - p1)
```

```python
class India:
    def capital(self):
        return "New Delhi"

    def language(self):
        return "Hindi / English"

    def type(self):
        return "Developing / Democratic Republic"


class USA:
    def capital(self):
        return "Washington, D.C."

    def language(self):
        return "English"

    def type(self):
        return "Developed / Federal Republic"


# Function demonstrating polymorphism
def country_details(country_obj):
    print(f"Capital: {country_obj.capital()}")
    print(f"Language: {country_obj.language()}")
    print(f"Type: {country_obj.type()}\n")


# Testing
country_details(India())
country_details(USA())
```

```python
class Product:
    def __init__(self, name, price, discount_percent):
        self.name = name
        self.price = price
        self.discount_percent = discount_percent

    def calculate_bill(self):
        discount_amount = (self.price * self.discount_percent) / 100
        final_bill = self.price - discount_amount
        return discount_amount, final_bill


# Testing
item = Product("Laptop", 50000, 10)
disc, total = item.calculate_bill()
print(f"Product: {item.name}")
print(f"Discount: ${disc}")
print(f"Final Bill: ${total}")
```

```python
class ClassA:
    def __init__(self, val):
        self.val = val


class ClassB:
    def __init__(self, val):
        self.val = val


class ClassC:
    def __init__(self, a_obj, b_obj):
        self.a = a_obj
        self.b = b_obj

    def add(self):
        return self.a.val + self.b.val

    def multiply(self):
        return self.a.val * self.b.val


# Testing
objA = ClassA(5)
objB = ClassB(4)
objC = ClassC(objA, objB)
print("Addition (A + B):", objC.add())
print("Multiplication (A * B):", objC.multiply())
```

```python
class Student:
    def __init__(self, name):
        self.name = name


class Marks(Student):
    def __init__(self, name, marks_list):
        super().__init__(name)
        self.marks_list = marks_list

    def find_highest(self):
        return max(self.marks_list)


# Testing
student_marks = Marks("Alice", [82, 95, 78, 91])
print(f"Student: {student_marks.name}")
print(f"Highest Marks: {student_marks.find_highest()}")
```

```python
def recursive_login(attempt=1, max_attempts=3):
    if attempt > max_attempts:
        print("Login Failure: Maximum attempts exceeded.")
        return

    print(f"\nAttempt {attempt} of {max_attempts}")
    username = input("Enter username: ")
    password = input("Enter password: ")

    if username == "root" and password == "admin":
        print("Login Success")
    else:
        print("Login Failure. Try again.")
        recursive_login(attempt + 1, max_attempts)


# Uncomment to run
# recursive_login()
```
