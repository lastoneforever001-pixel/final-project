```python
# sqlite connect to database
import sqlite3
conn = sqlite3.connect("student.db")
cursor = conn.cursor()
```

```python
sql = '''
CREATE TABLE Student(
    id INTEGER PRIMARY KEY,
    name TEXT,
    age INTEGER,
    address TEXT
)
'''
cursor.execute(sql)
```

```python
sql='''
INSERT INTO STUDENT(name,age,address)
VALUES ('ram',20,'kathmandu')
'''
cursor.execute(sql)
conn.commit()
print("data saved succefully")
```

```python
# read data
sql = '''
SELECT*FROM student
'''
cursor.execute(sql)
resault=cursor.fetchall()
print(resault)
```

```python
sql = '''
update student SET name = "HARI" WHERE id=2
'''
cursor.execute(sql)
conn.commit()
print("data saved successfully")
```

```python
sql='''
DELETE FROM STUDENT WHERE id = 2
'''
cursor.execute(sql)
conn.commit()
print("data saved succefully")
```

```python
# filter data
sql = '''
SELECT * FROM STUDENT WHERE name like "o%"
'''
cursor.execute(sql)
resault=cursor.fetchall()
print(resault)
```

```python
# insert commit database 
a=input("enter your name ")
b=int(input("enter your age :"))
c= input("enter your address ")
sql = f'''
INSERT INTO Student(name,age,address)
VALUES('{a}',{b},"{c}")


'''
cursor.execute(sql)
conn.commit()
print("data saved successfully!")
```
