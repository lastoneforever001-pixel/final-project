```python
class cal:
    def volume(self,l,b,h):
        print(l*b*h)
obj= cal()
obj.volume(10,10,10)
```

```python
class cal:
    def volume(self,l,b,h):
        print(l*b*h)
    def area(self,l,b):
        print(l*b)
obj = cal()
obj.volume(10,10,10)
obj.area(10,10)
    
```

```python
class cal:
    def __init__(self,l,b,h):
        self.l=l
        self.b=b
        self.h=h
    def volume(self):
         print(self.l*self.b*self.h)
    def area(self):
         print(self.l*self.b)
obj = cal(10,10,10)
obj.volume()
obj.area()
```

## wap to print circumference of circle to call the instant method

```python
class circle():
    pi = 3.14
    def __init__(self,radius):
        self.radius=radius

    def get_circumference(self):
        print(f"the circumference of a circle {2*self.pi*self.radius}")
obj=circle(4)
obj.get_circumference()
        
```

```python

class Age:
    def __init__(self):
        self.age = 40


class Address:
    def __init__(self):
        self.address = "KTM"


class Info(Age, Address):
    def __init__(self):
        self.name = "Ram"
        Age.__init__(self)
        Address.__init__(self)

    def info(self):
        print(f"Your name, age, address: {self.name}\t{self.age}\t{self.address}")


obj = Info()
obj.info()

```

```python
class age:
    def __init__(self,age):
        self.age=age
    def age(self):
        print(self.age)
class address:
    def __init__(self,add):
        self.add=add
class info(age,address):
    def __init__(self,name,age,add):
        self.name=name
        age.__init__(self,age)
        address.__init__(self,add)
    def info(self):
        print(f"hello: {self.name}\t{self.age}\t{self.add}")

obj=info("rahul",23,"ktm")
obj.info()
        
```
