# filter function 

```python
l= [1,2,3,4,5,6,7,8,9]
even=list(filter(lambda x:x%2==0,l))
print (even)
```

# map function 

```python
l=[1,2,3,4,5,6,7,8,9]
even=list(map(lambda x:x*10,l))
print (even)
```

#### reduce functiom

```python
ab= [1,2,3,4,5,6,7,8,9]
from functools import reduce
sum1 = reduce(lambda a,b:a+b,ab)
la = reduce(lambda a,b:a if a>b else b,ab)
sec = reduce(lambda a,b:a if a>b else b,[x for x in ab if x !=la ])
print (sec)
```
