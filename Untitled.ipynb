```python
def main():
    # 1. Input the number of students
    try:
        n = int(input("Enter the number of students: "))
    except ValueError:
        print("Please enter a valid integer for the number of students.")
        return

    students_data = {}
    subjects = ["Math", "Science", "English", "Computer"]

    # 2. Input marks and store in a nested dictionary
    print("\n--- Enter Student Details ---")
    for i in range(n):
        name = input(f"\nEnter the name of student {i+1}: ").strip()
        marks = {}
        for subject in subjects:
            while True:
                try:
                    score = float(input(f"  Enter marks for {subject}: "))
                    if 0 <= score <= 100:
                        marks[subject] = score
                        break
                    else:
                        print("  Marks should be between 0 and 100. Try again.")
                except ValueError:
                    print("  Invalid input. Please enter a number.")
        students_data[name] = marks

    # Variables to track pass/fail counts and the topper
    passed_count = 0
    failed_count = 0
    topper_name = None
    highest_average = -1.0

    print("\n--- Results ---")

    # 3. Process the data
    for name, marks in students_data.items():
        # Check if the student passed all subjects
        is_passed = all(score >= 40 for score in marks.values())

        if is_passed:
            passed_count += 1
            # Calculate average marks for the passed student
            avg_marks = sum(marks.values()) / len(marks)
            print(f"{name}: Passed | Average Marks = {avg_marks:.2f}")

            # Check if this student has the highest average so far
            if avg_marks > highest_average:
                highest_average = avg_marks
                topper_name = name
        else:
            failed_count += 1
            print(f"{name}: Failed")

    # 4. Display final summary and topper information
    print("\n--- Summary ---")
    print(f"Total Passed Students: {passed_count}")
    print(f"Total Failed Students: {failed_count}")

    if passed_count > 0:
        print(f"Topper: {topper_name} with the highest average of {highest_average:.2f}")
    else:
        print("No topper (all students failed).")


if __name__ == "__main__":
    main()
def main():
    # 1. Input the number of students
    try:
        n = int(input("Enter the number of students: "))
    except ValueError:
        print("Please enter a valid integer for the number of students.")
        return

    students_data = {}
    subjects = ["Math", "Science", "English", "Computer"]

    # 2. Input marks and store in a nested dictionary
    print("\n--- Enter Student Details ---")
    for i in range(n):
        name = input(f"\nEnter the name of student {i+1}: ").strip()
        marks = {}
        for subject in subjects:
            while True:
                try:
                    score = float(input(f"  Enter marks for {subject}: "))
                    if 0 <= score <= 100:
                        marks[subject] = score
                        break
                    else:
                        print("  Marks should be between 0 and 100. Try again.")
                except ValueError:
                    print("  Invalid input. Please enter a number.")
        students_data[name] = marks

    # Variables to track pass/fail counts and the topper
    passed_count = 0
    failed_count = 0
    topper_name = None
    highest_average = -1.0

    print("\n--- Results ---")

    # 3. Process the data
    for name, marks in students_data.items():
        # Check if the student passed all subjects
        is_passed = all(score >= 40 for score in marks.values())

        if is_passed:
            passed_count += 1
            # Calculate average marks for the passed student
            avg_marks = sum(marks.values()) / len(marks)
            print(f"{name}: Passed | Average Marks = {avg_marks:.2f}")

            # Check if this student has the highest average so far
            if avg_marks > highest_average:
                highest_average = avg_marks
                topper_name = name
        else:
            failed_count += 1
            print(f"{name}: Failed")

    # 4. Display final summary and topper information
    print("\n--- Summary ---")
    print(f"Total Passed Students: {passed_count}")
    print(f"Total Failed Students: {failed_count}")

    if passed_count > 0:
        print(f"Topper: {topper_name} with the highest average of {highest_average:.2f}")
    else:
        print("No topper (all students failed).")


if __name__ == "__main__":
    main()

```

```python
data = {"a": 10, "b": 50, "c": 20}

# Initialize max and min with the first value in the dictionary
values = list(data.values())
max_val = values[0]
min_val = values[0]

# Loop through values to find max and min
for val in values:
    if val > max_val:
        max_val = val
    if val < min_val:
        min_val = val

difference = max_val - min_val

print(f"Maximum Value: {max_val}"
    )
print(f"Minimum Value: {min_val}")
print(f"Difference: {difference}")
```

## wap to create a student resault management system using nested functions.calculate n subjects totals percentage division and pass /fail status

```python
def student_result_system():
    # Outer function to manage student data input and processing
    print("=== Student Result Management System ===")
    
    # Input student details
    student_name = input("Enter student name: ")
    try:
        num_subjects = int(input("Enter the number of subjects: "))
        if num_subjects <= 0:
            print("Number of subjects must be greater than 0.")
            return
            
        marks = []
        for i in range(1, num_subjects + 1):
            mark = float(input(f"Enter marks for subject {i} (out of 100): "))
            if mark < 0 or mark > 100:
                print("Invalid marks! Marks should be between 0 and 100.")
                return
            marks.append(mark)
    except ValueError:
        print("Invalid input! Please enter numeric values.")
        return

    # 1. Nested function to calculate total marks
    def calculate_total():
        return sum(marks)

    # 2. Nested function to calculate percentage
    def calculate_percentage(total_marks):
        return (total_marks / (num_subjects * 100)) * 100

    # 3. Nested function to check pass/fail status (passing mark is 40 per subject)
    def check_status():
        for mark in marks:
            if mark < 40:
                return "Fail"
        return "Pass"

    # 4. Nested function to determine division based on percentage
    def determine_division(percentage, status):
        if status == "Fail":
            return "No Division (Failed)"
        elif percentage >= 80:
            return "Distinction"
        elif percentage >= 60:
            return "First Division"
        elif percentage >= 50:
            return "Second Division"
        elif percentage >= 40:
            return "Third Division"
        else:
            return "No Division"

    # Executive logic inside the outer function
    total = calculate_total()
    percentage = calculate_percentage(total)
    status = check_status()
    division = determine_division(percentage, status)

    # Displaying the report card
    print("\n================ REPORT CARD ================")
    print(f"Student Name : {student_name}")
    print(f"Total Marks  : {total:.2f} / {num_subjects * 100}")
    print(f"Percentage   : {percentage:.2f}%")
    print(f"Status       : {status}")
    print(f"Division     : {division}")
    print("=============================================")

# Call the system
student_result_system()
```
